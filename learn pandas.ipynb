# **PANADAS LIBRARY TUTORIAL**

```python
import pandas as pd

df = pd.read_csv('../sales.csv') # stores the 'sales.csv' file as a dataframe in pandas
```

## ***DATAFRAME FUNDAMENTALS***

```python
df.head() # the first 5 rows
df.tail() # the last 5 rows
df.info() # gives general info: # of rows and cols, non-null values, Pandas Basics types
df.describe() # gives a description
df.columns # gives a list of the columns
df.index # gives a range

df[['Category','Product','Price']].head # indexing by column
df.loc[10, 'Price'] # indexing by row
df.iloc[10, 3] # indexing by row
df.loc[df['Product'] == 'Microwave'] # indexing by row
```

## ***FILTERING DATA***

```python
df[df['Category'] == 'Electronics'] # filters all products in Electronics category
df[(df['Category'] == 'Electronics') & (df['Price'] > 200)] # another layer of filter

# | for or
# & for and
# ~ in front for "reverse"

df[df['Category'].str.endswith('s')]
df[df['Category'].isin(['Electronics','Appliances'])]
```

## ***UPDATING DATA***

```python
df.loc[df['Product'] == 'Microwave', 'Price'] = 67.99 # changes the price of Microwave to 67.99
df["Category"] = df["Category"].str.upper() # capitalizes all Category values

df.drop(39) # clears row 39
df.dropna() # drops the entire row of missing values
df.fillna({"Price": 99.99}, inplace = True) # replaces missing values to 99.99
df.rename(columns={'Product': 'Item_Name', 'Price': 'Unit_Price'}, inplace = True) # Rename 'Product' to 'Item_Name' and 'Price' to 'Unit_Price'

# inplace=True makes it temporary
```

## ***ANALYZING DATA P2***

```python
df["Country"].value_counts()  # Counts how many times each unique item appears (e.g., USA: 5, Canada: 3)
df["Country"].unique()        # Returns an array of every unique item name, with no duplicates
df["Country"].nunique()       # Returns the *number* of unique items (the count of .unique())
df["Country"].count()         # Counts the total number of non-missing (non-NaN) rows in the column
df["Country"].mean()          # Calculates the average value (only works if the column contains numbers)
df["Country"].median()        # Finds the exact middle value of the Pandas Basics when sorted in order
df["Country"].mode()          # Finds the most frequently occurring value in the column
df["Country"].std()           # Measures standard deviation (how spread out the numbers are from the average)
df["Country"].var()           # Measures variance (standard deviation squared; another way to see Pandas Basics spread)
df["Country"].min()           # Finds the lowest value (or alphabetically first text item)
df["Country"].max()           # Finds the highest value (or alphabetically last text item)
```

## ***EXPORTING FILE***

```python
# Saves your current DataFrame as a new file named 'cleaned_sales.csv'
df.to_csv("cleaned_sales.csv")

# Saves the file cleanly without adding an extra column for row numbers
df.to_csv("cleaned_sales.csv", index=False)
```
