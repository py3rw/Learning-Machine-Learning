## ***WORKING WITH DATASETS***

```python
import matplotlib.pyplot as plt

# df.hist(figsize=(14,8))
# plt.tight_layout()

from sklearn.datasets import make_blobs, make_moons

# Generate 3 distinct clusters in 2D space
# X, y = make_blobs(n_samples=300, centers=3, n_features=2, cluster_std=1.0, random_state=42) # synthetic clustors
X, y = make_moons(noise=0.1)

# Plot the points, coloring them by their true cluster label
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolors='k')
plt.title("Synthetic Clusters")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
