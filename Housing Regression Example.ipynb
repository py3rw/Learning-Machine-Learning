# ***EXAMPLE OF REGRESSION IN ML***
***Using fetch_california_housing dataset and linear regression, this model will predict the price of the house (1.6 -> 160k)***

## ***DATASET & IMPORTS***

```python
from sklearn.datasets import fetch_california_housing  # loads a dataset
from sklearn.preprocessing import StandardScaler # allows scaling
from sklearn.model_selection import train_test_split # splitting data
```

```python
fetch_california_housing() # checking info on dataset
```

```python
df = fetch_california_housing(as_frame=True).frame # creating a pandas dataframe of dataset
df # showing df
```

## ***SPLITTING DATA & PREPROCESSING***

```python
X, y = fetch_california_housing(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) # 20% testing, 80% training

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## ***IMPORTING CLASSIFICATION***

```python
from sklearn.linear_model import LinearRegression

reg = LinearRegression()

reg.fit(X_train_scaled, y_train)
```

## ***CHECKING SCORE***

```python
reg.score(X_test_scaled, y_test) # not the exact % accuracy, instead it's the R^2 score.
```

## ***PREDICTING INSTANCE***

```python
print("Predicted:", reg.predict([X_test_scaled[0]])[0])
print("Actual:", y_test[0])
```
