# ***SCIKIT-LEARN DATA SPLITS TUTORIAL***

```python
from sklearn.model_selection import train_test_split # splits the data into a training and testing portion
from sklearn.datasets import load_iris
```

## ***SETUP OF DATA SPLIT***

```python
data = load_iris()
X, y = data.data, data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) # 20% testing, 80% training
# X is the inputs/features, y is the output/answer (targets)
# len(x) -> 150: len(X_train) -> 120 instances, len(X_test) -> 30 instances (RANDOM SPLIT)
```

## ***DISTRIBUTION OF CLASSIFICATION***

```python
import numpy as np
import matplotlib.pyplot as plt

counts = np.bincount(y_train) # counts the frequency of each, e.g. [40, 38, 42]
positions = np.arange(3) # creates the position x-axis

# graphing
plt.bar(positions, counts)
plt.xticks(positions, data.target_names)
```

## ***BALANCING TEST AND TRAIN DATA***

```python
from sklearn.model_selection import StratifiedShuffleSplit as sss

split = sss(n_splits=1, test_size=0.2)

for train_idx, test_idx in split.split(X,y):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]
```
