# ***SCIKIT-LEARN PREPROCESSING TUTORIAL***

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler # two main scalers
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2)
```

## ***SCALING DATA***
***Prevents a singular feature from dominating just because of its larger numbers***

```python
scaler = StandardScaler()
# gets the mean then divide by the standard deviation for each feature
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler = MinMaxScaler()
# compresses the data to a value between 0 and 1
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## ***FEATURE ENCODING***

```python
from sklearn.datasets import fetch_openml # downloads public datasets from OpenML repo
from sklearn.preprocessing import OrdinalEncoder # changes text into number values bc all ML algorithms must use numbers
from sklearn.preprocessing import OneHotEncoder # splits the features into more features with 1s(yes) and 0s(no)
```

```python
data = fetch_openml('car', version=2, as_frame=True).frame
data
```

```python
# Ordinal Encoding
columns_to_encode = ['lug_boot', 'safety']

encoder = OrdinalEncoder(
    categories = [
    ['small', 'med', 'big'], # 0.0, 1.0, 2.0
    ['low', 'med', 'high'], # 0.0, 1.0, 2.0
    ]
)

data[columns_to_encode] = encoder.fit_transform(data[columns_to_encode])

data[columns_to_encode] = encoder.inverse_transform(data[columns_to_encode])
```

```python
data = fetch_openml('adult', version=1, as_frame=True).frame
data
```

```python
# One Hot Encoding
# splits the features

import pandas as pd
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False) # ignores unknown output, returns dense numpy array
encoded_values = encoder.fit_transform(data[['occupation', 'race']]) # fit: scans, transform: converts
new_cols = encoder.get_feature_names_out(['occupation','race']) # splits it into columns with 1s and 0s

df_encoded = pd.DataFrame(encoded_values, columns= new_cols, index=data.index) # pd.DataFrame(): converts into a pd dataframe, columns=new_cols: applies readable feature names, index=data.index: matches the row indices to the original df
data_final = pd.concat(
    [data.drop(columns=['occupation','race']), df_encoded], axis=1 # removes the old string-based features, pairs the remaining columns, axis=1: tells it to concat horizontally
)

data_final
```
