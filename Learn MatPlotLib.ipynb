# ***MATPLOTLIB LIBRARY TUTORIAL***

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd

%matplotlib inline
# displays plots inline on Jupyter notebooks
sns.set_theme(context="notebook", style="whitegrid", palette="deep")

# plt.style.use('seaborn') # Styles like 'ggplot', 'seaborn', etc
```

## ***GRAPHING LINE PLOTS***

```python
# Method 1: Simple Pyplot approach

x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.title("Simple Line Plot")
plt.xlabel("terrence")
plt.ylabel("situ")
plt.show()

# plt.show() displays the plot however this is automatic in Jupyter Notebooks
```

```python
# Method 2: Object-Oriented approach

fig, ax = plt.subplots() # creates a figure and axes object
ax.plot(x,y)
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('Line PLot (Object Oriented)')
plt.show()
```

```python
# Styles
x = np.linspace(0, 10, 100)
plt.figure(figsize=(10, 6))

plt.plot(x, x + 0, linestyle='-', linewidth=2.5, label="'-' or 'solid'") # solid
plt.plot(x, x + 2, linestyle='--', linewidth=2.5, label="'--' or 'dashed'") # dash
plt.plot(x, x + 4, linestyle='-.', linewidth=2.5, label="'-.' or 'dashdot'") # dash dots
plt.plot(x, x + 6, linestyle=':', linewidth=2.5, label="':' or 'dotted'") # dotted
plt.plot(x, x + 8, linestyle=(0, (5, 2, 1, 2)), linewidth=2.5, label="Custom: (0, (5, 2, 1, 2))") # custom pattern

plt.title("Matplotlib Line Styles Overview", fontsize=14, fontweight="bold")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend(title="Line Style Code", frameon=True, facecolor="white")
plt.show()
```

## ***GRAPHING SCATTER PLOT***

```python
np.random.seed(42)
x = np.random.randn(100)
y = 2*x+np.random.randn(100)*0.5 # Linear with noise

plt.figure(figsize=(8, 6))
plt.scatter(x,y,alpha=0.6) # alpha controls transparency
plt.grid(True,alpha=0.3)
plt.show()
```

```python
fig, ax = plt.subplots()

ax.scatter(
    x, y,
    s=100,            # SIZE: Single number OR array of numbers to scale points
    c='blue',         # COLOR: Name/hex code OR array mapped to a colormap (cmap)
    marker='o',       # SHAPE: 'o' circle, 's' square, '^' triangle, '*' star, 'x'
    alpha=0.5,        # TRANSPARENCY: 0.0 (invisible) to 1.0 (solid) - fixes overlap
    edgecolors='black'# BORDER: Outlines points so they stand out
)

plt.show()
```

## ***GRAPHING BAR CHARTS***

```python
# Simple Bar Chart
categories = ["A", "B", "C", "D", "E"]
values = [23, 45, 56, 78, 32]

plt.figure(figsize=(8, 6))
plt.bar(categories, values)  # Pass X (categories) and Y (values) here!

plt.title("Simple Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")

plt.show()
```

```python
# object-oriented bar chart

fig, ax = plt.subplots(figsize=(8, 6))
ax.bar(categories, values)
ax.set(title="Simple Bar Chart", xlabel="Categories", ylabel="Values")
plt.show()
```

```python
# Horizontal

categories = ["A", "B", "C", "D", "E"]
values = [23, 45, 56, 78, 32]

fig, ax = plt.subplots(figsize=(8, 4))

# Notice ax.barh instead of ax.bar!
ax.barh(categories, values, color="skyblue", edgecolor="black")

ax.set_title("Horizontal Bar Chart")
ax.set_xlabel("Values")
ax.set_ylabel("Categories")

plt.show()
```

```python
# Comparison Bar Chart

categories = ["A", "B", "C", "D", "E"]
group1 = [23, 45, 56, 78, 32]
group2 = [35, 20, 60, 45, 50]

x = np.arange(len(categories))  # Label locations [0, 1, 2, 3, 4]
width = 0.35                    # Width of each bar

fig, ax = plt.subplots(figsize=(8, 5))

# Offset the bars along the x-axis
ax.bar(x - width/2, group1, width, label="Group 1", color="teal")
ax.bar(x + width/2, group2, width, label="Group 2", color="coral")

ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.set_title("Grouped Bar Chart")
ax.legend()

plt.show()
```

```python
# Stacked Bar Chart

categories = ["A", "B", "C", "D", "E"]
group1 = [23, 45, 56, 78, 32]
group2 = [35, 20, 60, 45, 50]

fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(categories, group1, label="Group 1", color="#4C72B0")
# "bottom=group1" stacks group2 on top of group1
ax.bar(categories, group2, bottom=group1, label="Group 2", color="#DD8452")

ax.set_title("Stacked Bar Chart")
ax.legend()

plt.show()
```

```python
# Patterned/Hatched Bar Chart

categories = ["A", "B", "C", "D", "E"]
values = [23, 45, 56, 78, 32]
# Hatch patterns: '/', '\\', '|', '-', '+', 'x', 'o', '*', '.'
hatches = ["/", "\\", "|", "x", "*"]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(categories, values, color="lightgray", edgecolor="black", linewidth=1.5)

# Apply a different pattern to each bar
for bar, hatch in zip(bars, hatches):
    bar.set_hatch(hatch)

ax.set_title("Patterned (Hatched) Bar Chart")
plt.show()
```

## ***SUBPLOTS***

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 8)) # creates 4 grids

# Top-Left [0, 0]
ax[0, 0].plot([1, 2, 3], [4, 5, 6], color="blue")
ax[0, 0].set_title("Top Left")

# Top-Right [0, 1]
ax[0, 1].bar(["A", "B"], [10, 20], color="orange")
ax[0, 1].set_title("Top Right")

# Bottom-Left [1, 0]
ax[1, 0].scatter([1, 2, 3], [3, 2, 1], color="green")
ax[1, 0].set_title("Bottom Left")

# Bottom-Right [1, 1]
ax[1, 1].hist([1, 2, 2, 3, 3, 3, 4], color="purple")
ax[1, 1].set_title("Bottom Right")

plt.tight_layout()
plt.show()
```

## ***MATPLOTLIB WITH PANDAS***

```python
# Average Bar Chart for Sales
df = pd.read_csv('../sales.csv')

fig, ax = plt.subplots(figsize=(8,6))

sns.barplot(
    data=df,
    x = 'Category',
    y = 'Price',
    estimator="mean",
    hue= 'Category',
    legend= True,
    ax = ax,
    palette = "Blues_d"
)

ax.set_title("Average Price per Category", fontsize=14, fontweight="bold")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Most Expensive Items and Order_ID
df = pd.read_csv('../sales.csv')

fig, ax = plt.subplots(figsize=(8,6))

sns.scatterplot(
    data=df,
    x = 'Order_ID',
    y = 'Price',
    legend = True,
    ax = ax,
)

ax.set_title("Price vs Order Number", fontsize=14, fontweight="bold")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```
