# ***NUMPY LIBRARY TUTORIAL***

```python
import numpy as np

# Create a basic 1D array
arr = np.array([1, 2, 3, 4, 5])
arr1 = np.array([2, 4, 6, 8, 10])
arr2 = np.array([3, 6, 9, 12, 15])

# Create a 2D matrix
matrix = np.array([[1, 2, 3], [4, 5, 6]])
```

## ***ARRAY CREATION & PROPERTIES***

```python
arr.ndim  # Returns number of dimensions (e.g., 1D, 2D)
arr.shape  # Returns dimensions as a tuple (rows, cols)
arr.size  # Returns total number of elements in array
arr.dtype  # Returns data type of elements (e.g., int64, float64)

np.zeros((3, 3))  # Creates a 3x3 array filled with 0s
np.ones((2, 4))  # Creates a 2x4 array filled with 1s
np.full((2, 2), 7)  # Creates a 2x2 array filled with custom value (7)
np.arange(0, 10, 2)  # Array from 0 to 9 with step size of 2
np.linspace(0, 1, 5)  # 5 evenly spaced numbers from 0 to 1
np.random.rand(3, 3)  # 3x3 matrix with random floats between 0 and 1
```

## ***INDEXING & SLICING***

```python
arr[0]  # Gets element at index 0
arr[-1]  # Gets last element
matrix[0, 1]  # Gets row 0, column 1 element in a 2D matrix

arr[1:4]  # Slices elements from index 1 to 3
matrix[:, 0]  # Gets all rows, but only column 0
matrix[0, :]  # Gets row 0, across all columns

# Filtering Data (Boolean Masking)
arr[arr > 2]  # Filters array to elements greater than 2
```

## ***ARRAY OPERATIONS & MATH***

```python
arr + 5  # Adds 5 to every element (Broadcasting)
arr * 2  # Multiplies every element by 2
arr1 + arr2  # Element-wise addition between two arrays

np.dot(matrix1, matrix2)  # Matrix multiplication (Dot product)
matrix1 @ matrix2  # Shorthand for matrix multiplication

np.sin(arr)  # Calculates sine for each element
np.exp(arr)  # Exponential (e^x) for each element
np.sqrt(arr)  # Square root of each element
```

## ***STATISTICAL ANALYSIS***

```python
arr.mean()  # Calculates the average of all elements
arr.std()  # Calculates standard deviation
arr.sum()  # Sums all elements together
arr.min()  # Finds lowest value
arr.max()  # Finds highest value

np.median(arr)  # Calculates the median value
arr.argmin()  # Returns index of the minimum value
arr.argmax()  # Returns index of the maximum value

matrix.sum(axis=0)  # Sums along columns
matrix.sum(axis=1)  # Sums along rows
```

## ***RESHAPING & MANIPULATION***

```python
arr.reshape(2, 5)  # Changes array shape to 2 rows, 5 cols (must match size)
matrix.flatten()  # Converts multi-dimensional array into 1D array
matrix.T  # Transposes matrix (swaps rows and columns)

np.concatenate((arr1, arr2))  # Joins two arrays together
np.vstack((arr1, arr2))  # Stacks arrays vertically (row-wise)
np.hstack((arr1, arr2))  # Stacks arrays horizontally (col-wise)
```
